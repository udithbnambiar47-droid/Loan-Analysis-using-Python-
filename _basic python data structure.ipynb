```python
#data structure
Python provides built-in data structures to store and manage data efficiently.
```

```python
#list
Ordered, mutable (can change)
Allows duplicate values
```

```python
numbers = [10, 20, 30, 40]

numbers.append(50)   # Add
numbers.remove(20)   # Remove
print(numbers)
```

```python
#tuple
Ordered, immutable (cannot change)
```

```python
data = (10,20,30)
print(data[0])
```

```python
👉 Use when: Data should NOT change (safe & faster)
```

```python
#dictionary
Key-value pairs
Fast lookup
```

```python
student= {
    "name":"udith",
    "age":20
}
print(student["name"])
```

```python
use when you want labeled data (like data base records)
```

```python
#Set
Unordered, no duplicates
```

```python
nums = {1,2,2,3,3,4,}
print(nums)
```

```python
use when you want unique value
```

```python
#String (Special Data Structure)
Sequence of characters
```

```python
text = "python"
print (text[0])
```

```python
#Stack (LIFO)
Last In First Out
```

```python
stack = []
stack.append(10)
stack.append(20)
stack.pop()
```

```python
#Queue (FIFO)
First In First Out
```

```python
from collections import deque
queue = deque()
queue.append(10)
queue.append(20)
queue.popleft()
```
