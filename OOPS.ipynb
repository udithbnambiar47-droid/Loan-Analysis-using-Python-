```python
#Inheritance
#One class can use properties of another class.
```

```python
class Animal:
    def speak(self):
        print("animal speak")

class Dog(Animal):   
    pass

d = Dog()
d.speak()
```

```python
#Polymorphism
#Same function name, different behavior.
```

```python
class dog:
    def sound(self):
        print("bark")

class cat:
    def sound(self):
        print("meow")
```

```python
d = dog()
c = cat()

d.sound()
c.sound()
```

```python
#if-else → You decide manually
#Polymorphism → Python decides automatically
#Both classes have the same method name → sound()
But they behave differently
Dog → Bark 🐶
Cat → Meow 🐱

👉 This is called Polymorphism (same name, different behavior)#Same function name, different behavior.
```

```python
#Abstraction
#Hiding internal details, showing only important parts.
#Abstraction in Python
#We use the abc module to create abstract classes
```

```python
from abc import ABC, abstractmethod

class payment(ABC):
    @abstractmethod
    def pay(self):
        pass

class UPI(payment):
    def pay(self):
        print("pay using UPI")

class card(payment):
    def pay(self):
        print("pay using card")

p = UPI()
p.pay()
        
```
