# Good Loan Applications 

```python
# Good Loan = Fully Paid
good_loan = df[df["loan_status"] == "Fully Paid"]

# 1. Good Loan Applications
good_loan_applications = good_loan["id"].nunique()

# Total applications
total_loan_applications = df["id"].nunique()

# 2. Good Loan Application Percentage
good_loan_percentage = (
    good_loan_applications / total_loan_applications
) * 100

# 3. Good Loan Funded Amount
good_loan_funded_amount = good_loan["loan_amount"].sum()

# 4. Good Loan Total Received Amount
good_loan_total_received = good_loan["total_payment"].sum()


# Display results
print("========== GOOD LOAN ==========")

print("Good Loan Applications:",
      good_loan_applications)

print("Good Loan Application Percentage:",
      f"{good_loan_percentage:.2f}%")

print("Good Loan Funded Amount:",
      f"₹{good_loan_funded_amount:,.0f}")

print("Good Loan Total Received Amount:",
      f"₹{good_loan_total_received:,.0f}")
```

# Bad Loan Applications

```python

# Bad Loan = Charged Off
bad_loan = df[df["loan_status"] == "Charged Off"]

# Total Loan Applications
total_loan_applications = df["id"].nunique()

# 1. Bad Loan Applications
bad_loan_applications = bad_loan["id"].nunique()

# 2. Bad Loan Application Percentage
bad_loan_percentage = (
    bad_loan_applications / total_loan_applications
) * 100

# 3. Bad Loan Funded Amount
bad_loan_funded_amount = bad_loan["loan_amount"].sum()

# 4. Bad Loan Total Received Amount
bad_loan_total_received = bad_loan["total_payment"].sum()


# Display
print("========== BAD LOAN ==========")

print(
    "Bad Loan Application Percentage:",
    f"{bad_loan_percentage:.2f}%"
)

print(
    "Bad Loan Applications:",
    bad_loan_applications
)

print(
    "Bad Loan Funded Amount:",
    f"₹{bad_loan_funded_amount:,.0f}"
)

print(
    "Bad Loan Total Received Amount:",
    f"₹{bad_loan_total_received:,.0f}"
)
```
