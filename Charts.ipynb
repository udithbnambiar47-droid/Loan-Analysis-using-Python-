```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import warnings
import plotly.express as px
```

```python
import pandas as pd

df = pd.read_excel("C:/Users/Administrator/Downloads/financial_loan (6).xlsx")  

print(df.head())
```

```python
df['issue_date'] = pd.to_datetime(df['issue_date'])
```

```python
df['month_year'] = df['issue_date'].dt.to_period('M')
```

```python
monthly_trend = df.groupby('month_year').size()
```

```python
monthly_trend.index = monthly_trend.index.astype(str)
```

# Line Chart

```python
import matplotlib.pyplot as plt

plt.figure()
monthly_trend.plot(kind='line')

plt.title("Monthly Loan Application Trend")
plt.xlabel("Month")
plt.ylabel("Number of Applications")
plt.xticks(rotation=45)

plt.show()
```

# Area chart

```python
plt.figure()
monthly_trend.plot(kind='area')

plt.title("Monthly Loan Application Trend (Area Chart)")
plt.xlabel("Month")
plt.ylabel("Number of Applications")
plt.xticks(rotation=45)

plt.show()
```
