```python
try:
    a = int(input())
    print(10 / a)
except ValueError:
    print("invalid input")
except ZeroDivisionError:
    print("cannot divide by zero")
```

```python
#else and finally
```

```python
try :
    a =int(input())
    print(10 / a)
except:
    print("error")
else:
    print("no error")
finally:
    print("always run")
```
