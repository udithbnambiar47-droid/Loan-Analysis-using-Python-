# Total Funded Amount

**Understanding the total amount of funds disbursed as loans is crucial. We also want to keep an eye on the MTD Total Funded Amount metric.**


```python
# Total Funded Amount
total_funded_amount = df["loan_amount"].sum()

# Latest available issue date
latest_date = df["issue_date"].max()

# MTD data
mtd_data = df[
    (df["issue_date"].dt.year == latest_date.year) &
    (df["issue_date"].dt.month == latest_date.month)
]

# MTD Total Funded Amount
mtd_funded_amount = mtd_data["loan_amount"].sum()

# Results
print("Total Funded Amount: ₹{:,.0f}".format(total_funded_amount))
print("Latest Issue Date:", latest_date)
print("MTD Total Funded Amount: ₹{:,.0f}".format(mtd_funded_amount))
```

# Total Amount Received

**Tracking the total amount received from borrowers is essential for assessing the bank's cash flow and loan repayment. We should analyse the Month-to-Date (MTD) Total Amount Receive.**


```python
# Total Amount Received
total_amount_received = df["total_payment"].sum()

# Latest issue date
latest_date = df["issue_date"].max()

# MTD data
mtd_data = df[
    (df["issue_date"].dt.year == latest_date.year) &
    (df["issue_date"].dt.month == latest_date.month)
]

# MTD Total Amount Received
mtd_amount_received = mtd_data["total_payment"].sum()

# Results
print(f"Total Amount Received: ₹{total_amount_received:,.0f}")
print(f"Latest Issue Date: {latest_date}")
print(f"MTD Total Amount Received: ₹{mtd_amount_received:,.0f}")
```
