```python
def count_down(n):
    if n==0:
        return
    print(n)
    count_down(n-1)
    
count_down(5)
```

```python
def factorial(n):
    if n ==1:
        return 1
    return n * factorial(n-1)
print(factorial(5))
```

```python
def compound_interest(P,r,n):
    if n == 0:
        return P
    return compound_interest(P*(1+r),r,n-1)

print(compound_interest(1000,0.1,2))
```
